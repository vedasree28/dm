```python
import pandas as pd
data1 = {
'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
'Age': [20, 25, None, 30, 25],
'Salary': [30000, 40000, 35000, None, 40000],
'City': ['Hyderabad', 'Delhi', 'Mumbai', 'Hyderabad', None]
}
df1 = pd.DataFrame(data1)
print("Original Data:")
print(df1)

df1 = df1.drop_duplicates()
df1['Age'] = df1['Age'].fillna(df1['Age'].mean())
df1['Salary'] = df1['Salary'].fillna(df1['Salary'].mean())
df1['City'] = df1['City'].fillna(df1['City'].mode()[0])
print("\nAfter Data Cleaning:")
print(df1)

df1['Age'] = (df1['Age'] - df1['Age'].min()) / (df1['Age'].max() - df1['Age'].min())
df1['Salary'] = (df1['Salary'] - df1['Salary'].min()) / (df1['Salary'].max() - df1['Salary'].min())
print("\nAfter Normalization:")
print(df1)

data2 = {
'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
'Department': ['IT', 'HR', 'AI', 'IT', 'Finance']
}

df2 = pd.DataFrame(data2)
integrated_data = pd.merge(df1, df2, on='Name')
print("\nAfter Data Integration:")
print(integrated_data)
```

```python
import pandas as pd
data = {
'ID': [1, 2, 3, 4, 5, 6, 7, 8],
'Name': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
'Age': [20, 21, 22, 23, 24, 25, 26, 27],
'Marks': [85, 90, 78, 88, 92, 75, 80, 95]
}
df = pd.DataFrame(data)
print("Original Dataset:")
print(df)
# HORIZONTAL PARTITIONING - divide rows
horizontal_p1 = df.iloc[:4]
horizontal_p2 = df.iloc[4:]
print("\nHorizontal Partition 1:")
print(horizontal_p1)
print("\nHorizontal Partition 2:")
print(horizontal_p2)
# VERTICAL PARTITIONING - divide columns
vertical_p1 = df[['ID', 'Name']]
vertical_p2 = df[['ID', 'Age', 'Marks']]
print("\nVertical Partition 1:")
print(vertical_p1)
print("\nVertical Partition 2:")
print(vertical_p2)
# ROUND ROBIN PARTITIONING
n = 2
round_robin_p1 = df.iloc[0::n]
round_robin_p2 = df.iloc[1::n]
print("\nRound Robin Partition 1:")
print(round_robin_p1)
print("\nRound Robin Partition 2:")
print(round_robin_p2)
# HASH-BASED PARTITIONING
hash_p1 = df[df['ID'] % 2 == 0]
hash_p2 = df[df['ID'] % 2 != 0]
print("\nHash Partition 1 (ID % 2 = 0):")
print(hash_p1)
print("\nHash Partition 2 (ID % 2 != 0):")
print(hash_p2)
```

```python
import pandas as pd
# STAR SCHEMA: one fact table connected to dimension tables
fact_sales = pd.DataFrame({
'Product_ID': [1, 2, 3],
'Customer_ID': [101, 102, 103],
'Sales': [500, 700, 900]
})
dim_product = pd.DataFrame({
'Product_ID': [1, 2, 3],
'Product': ['Laptop', 'Phone', 'Tablet']
})
dim_customer = pd.DataFrame({
'Customer_ID': [101, 102, 103],
'Customer': ['A', 'B', 'C']
})

print("STAR SCHEMA")
print("Fact Sales:")
print(fact_sales)
print("Product Dimension:")
print(dim_product)
print("Customer Dimension:")
print(dim_customer)
# SNOWFLAKE SCHEMA: normalized dimension tables
dim_product = pd.DataFrame({
'Product_ID': [1, 2],
'Product': ['Laptop', 'Phone'],
'Category_ID': [10, 20]
})
dim_category = pd.DataFrame({
'Category_ID': [10, 20],
'Category': ['Computer', 'Mobile']
})

print("\nSNOWFLAKE SCHEMA")
print(dim_product)
print(dim_category)
snowflake = pd.merge(dim_product, dim_category, on='Category_ID')
print("After Joining:")
print(snowflake)
# FACT CONSTELLATION: multiple fact tables share dimensions
fact_sales = pd.DataFrame({
'Product_ID': [1, 2],
'Sales': [500, 700]
})
fact_returns = pd.DataFrame({
'Product_ID': [1, 2],
'Returns': [20, 30]
})
dim_product = pd.DataFrame({
'Product_ID': [1, 2],
'Product': ['Laptop', 'Phone']
})

print("\nFACT CONSTELLATION")
print("Sales Fact:")
print(fact_sales)
print("Returns Fact:")
print(fact_returns)
print("Shared Product Dimension:")
print(dim_product)
```

```python
import pandas as pd
data = {
'Year': [2024, 2024, 2025, 2025],
'City': ['Hyderabad', 'Delhi', 'Hyderabad', 'Delhi'],
'Product': ['Laptop', 'Mobile', 'Laptop', 'Mobile'],
'Sales': [50000, 30000, 60000, 40000]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
# ROLL-UP: summarize data at a higher level
rollup = df.groupby('Year')['Sales'].sum()
print("\n1. ROLL-UP - Total Sales by Year:")
print(rollup)
# DRILL-DOWN: view more detailed data
drilldown = df.groupby(['Year', 'City'])['Sales'].sum()
print("\n2. DRILL-DOWN - Sales by Year and City:")
print(drilldown)
# SLICE: select one value of a dimension
slice_data = df[df['Year'] == 2025]
print("\n3. SLICE - Sales for 2025:")
print(slice_data)
# DICE: select a sub-cube using multiple conditions
dice_data = df[(df['City'] == 'Hyderabad') & (df['Product'] == 'Laptop')]
print("\n4. DICE - Hyderabad Laptop Sales:")
print(dice_data)
# PIVOT: rotate dimensions for a different view
pivot = pd.pivot_table(df, values='Sales', index='City',
columns='Year', aggfunc='sum')
print("\n5. PIVOT - City vs Year:")
print(pivot)
```

```python
import pandas as pd
data = {
'Age': [20, 22, 25, 35, 38, 42, 45, 50],
'City': ['Hyderabad', 'Hyderabad', 'Delhi', 'Delhi',
'Mumbai', 'Mumbai', 'Chennai', 'Chennai'],
'Salary': [25000, 30000, 40000, 45000, 55000, 60000, 70000, 75000]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
# GENERALIZE AGE
def age_group(age):
    if age < 30:
        return 'Young'
    elif age < 40:
        return 'Middle'
    else:
        return 'Senior'
df['Age_Group'] = df['Age'].apply(age_group)
# GENERALIZE CITY
def city_group(city):
    if city in ['Hyderabad', 'Delhi']:
        return 'North/South Central'
    else:
        return 'Other Cities'
df['City_Group'] = df['City'].apply(city_group)
# GENERALIZE SALARY
def salary_group(salary):
    if salary < 40000:
        return 'Low'
    elif salary < 60000:
        return 'Medium'
    else:
        return 'High'
df['Salary_Group'] = df['Salary'].apply(salary_group)
# COUNT GENERALIZED TUPLES
result = df.groupby(
['Age_Group', 'City_Group', 'Salary_Group']
).size().reset_index(name='Count')
print("\nGeneralized Data:")
print(result)
```

```python
from itertools import combinations
transactions = [
['Milk', 'Bread'],
['Milk', 'Bread'],
['Milk', 'Egg'],
['Bread', 'Egg'],
['Milk', 'Bread']
]
min_support = 2
def support(itemset):
    count = 0
    for transaction in transactions:
        if set(itemset).issubset(set(transaction)):
            count += 1
    return count
items = sorted(set(item for t in transactions for item in t))
for k in range(1, len(items) + 1):
    print("\n", k, "-itemsets:")
    for itemset in combinations(items, k):
        count = support(itemset)
        if count >= min_support:
            print(itemset, "Support =", count / len(transactions))
```

```python
from collections import Counter
from itertools import combinations

T = [
    ['Milk', 'Bread'],
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter']
]

min_sup = 2

items = Counter(i for t in T for i in t)

items = [i for i in items if items[i] >= min_sup]

print("Frequent Itemsets:")

for n in range(1, len(items) + 1):
    count = Counter(
        x for t in T
        for x in combinations([i for i in t if i in items], n)
    )

    for x, c in count.items():
        if c >= min_sup:
            print(x, ":", c)
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
data = {
'Age': [20, 25, 30, 35, 40, 45],
'Income': [20, 25, 30, 40, 50, 60],
'Buy': ['No', 'No', 'Yes', 'Yes', 'Yes', 'Yes']
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
df['Buy'] = df['Buy'].map({'No': 0, 'Yes': 1})
X = df[['Age', 'Income']]
y = df['Buy']
model = DecisionTreeClassifier(
criterion='entropy',
max_depth=2,
random_state=1
)
model.fit(X, y)
prediction = model.predict(pd.DataFrame([[28, 28]],
columns=['Age', 'Income']))
print("\nPrediction:", "YES" if prediction[0] == 1 else "NO")
plot_tree(model, feature_names=['Age', 'Income'],
class_names=['No', 'Yes'], filled=True)
plt.show()
```

```python
import pandas as pd
import math
data = {
'Outlook': ['Sunny', 'Sunny', 'Rainy', 'Rainy', 'Sunny', 'Rainy'],
'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No']
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)
# ENTROPY FUNCTION
def entropy(series):
    counts = series.value_counts()
    total = len(series)
    result = 0
    for count in counts:
        p = count / total
        result -= p * math.log2(p)
    return result
# TOTAL ENTROPY
total_entropy = entropy(df['Play'])
print("\nTotal Entropy:")
print(round(total_entropy, 3))
# WEIGHTED ENTROPY
weighted_entropy = 0
for value in df['Outlook'].unique():
    subset = df[df['Outlook'] == value]
    e = entropy(subset['Play'])
    weight = len(subset) / len(df)
    weighted_entropy += weight * e
    print("\n", value)
    print("Entropy =", round(e, 3))
# INFORMATION GAIN
information_gain = total_entropy - weighted_entropy
print("\nWeighted Entropy:")
print(round(weighted_entropy, 3))
print("\nInformation Gain:")
print(round(information_gain, 3))
```

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder
data = {
'Weather': ['Sunny', 'Sunny', 'Rainy', 'Rainy', 'Overcast',
'Sunny', 'Overcast', 'Rainy'],
'Temperature': [30, 28, 20, 18, 25, 32, 24, 19],
'Play': ['No', 'No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'Yes']
}
df = pd.DataFrame(data)
# Convert categorical weather values into numbers
encoder = LabelEncoder()
df['Weather'] = encoder.fit_transform(df['Weather'])
df['Play'] = df['Play'].map({'No': 0, 'Yes': 1})
X = df[['Weather', 'Temperature']]
y = df['Play']
model = GaussianNB()
model.fit(X, y)
# Predict for a new sample: Sunny, temperature 29
weather_value = encoder.transform(['Sunny'])[0]
sample = pd.DataFrame([[weather_value, 29]],
columns=['Weather', 'Temperature'])
prediction = model.predict(sample)
print("Predicted Class:", "Yes" if prediction[0] == 1 else "No")
print("Class Probabilities:", model.predict_proba(sample))
```

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
data = {
'Height': [150, 160, 170, 180, 155, 175],
'Weight': [50, 55, 65, 80, 52, 75],
'Class': ['A', 'A', 'B', 'B', 'A', 'B']
}
df = pd.DataFrame(data)
X = df[['Height', 'Weight']]
y = df['Class']
# Scale features so that distance is not dominated by one column
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_scaled, y)
sample = pd.DataFrame([[165, 60]], columns=['Height', 'Weight'])
sample_scaled = scaler.transform(sample)
prediction = model.predict(sample_scaled)
print("Predicted Class:", prediction[0])
print("Nearest Neighbour Indices:", model.kneighbors(sample_scaled)[1])
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
data = {
'Age': [20, 22, 25, 27, 45, 47, 50, 52],
'Income': [20, 22, 25, 27, 45, 47, 50, 52]
}
df = pd.DataFrame(data)
model = KMeans(n_clusters=2, random_state=1, n_init=10)
model.fit(df)
df['Cluster'] = model.labels_
print("Clustered Data:")
print(df)
print("\nCluster Centers:")
print(model.cluster_centers_)
plt.scatter(df['Age'], df['Income'], c=df['Cluster'])
plt.xlabel('Age')
plt.ylabel('Income')
plt.title('K-Means Clustering')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import Birch
data = {
'X': [1, 2, 3, 8, 9, 10, 20, 21, 22],
'Y': [1, 2, 2, 8, 9, 8, 20, 21, 20]
}
df = pd.DataFrame(data)
model = Birch(n_clusters=3, threshold=0.5, branching_factor=50)
model.fit(df)
df['Cluster'] = model.predict(df)
print("BIRCH Clustered Data:")
print(df)
plt.scatter(df['X'], df['Y'], c=df['Cluster'])
plt.xlabel('X')
plt.ylabel('Y')
plt.title('BIRCH Clustering')
plt.show()
```

```python
# Implementation of PAM Clustering Algorithm

# Step 1: Import required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn_extra.cluster import KMedoids

# Step 2: Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# Step 3: Display dataset information
print("Dataset Shape:", X.shape)
print("First 5 Records:")
print(X[:5])

# Step 4: Create PAM model
model = KMedoids(n_clusters=3, method='pam',
                 random_state=42)

# Step 5: Train the model
model.fit(X)

# Step 6: Predict cluster labels
labels = model.labels_

# Step 7: Display cluster labels
print("Cluster Labels:")
print(labels)

# Step 8: Display medoid points
print("Medoid Points:")
print(model.cluster_centers_)

# Step 9: Visualize the clusters
plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1],
            c=labels, cmap='rainbow')

# Step 10: Display graph details
plt.title("PAM Clustering on Iris Dataset")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```
